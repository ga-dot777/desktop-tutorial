```python
import requests
import json

# URL до сирого JSON-файлу відкритого трекера Epic Free Games
url = "https://raw.githubusercontent.com/josephmate/EpicFreeGamesList/main/epic_free_games.json"

print("Downloading the Epic Games Store free games database...")

response = requests.get(url)

if response.status_code == 200:
    games = response.json()
    print(f"✅ Successfully downloaded {len(games)} games!\n")

    print("Example entry:")
    print(json.dumps(games[0], indent=2, ensure_ascii=False))
else:
    print(f"❌ Download error: {response.status_code}")
```

```python
import pandas as pd
import json

df = pd.DataFrame(games)

print("=== Initial table ===")
print(f"Size: {df.shape[0]} rows, {df.shape[1]} columns")

columns_to_keep = ['gameTitle', 'freeDate', 'epicRating', 'metacriticScore', 'steamDBRating']
df_cleaned = df[columns_to_keep].copy()

df_cleaned['freeDate'] = pd.to_datetime(df_cleaned['freeDate'])

import numpy as np
df_cleaned.replace(0, np.nan, inplace=True)

print("\n=== Cleaned table ===")
print(df_cleaned.head())


csv_filename = "epic_free_games_cleaned.csv"
df_cleaned.to_csv(csv_filename, index=False)

print(f"\n✅ Data successfully cleaned and saved to file: {csv_filename}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_cleaned = pd.read_csv("epic_free_games_cleaned.csv")
df_cleaned['freeDate'] = pd.to_datetime(df_cleaned['freeDate'])

df_cleaned['Year'] = df_cleaned['freeDate'].dt.year

print("=== 1. BASE RATING STATISTICS ===")
# Convert epicRating (5-point scale) to a 100-point scale for easier comparison
df_cleaned['epicRating_100'] = df_cleaned['epicRating'] * 20
print(df_cleaned[['epicRating_100', 'metacriticScore', 'steamDBRating']].describe().round(2))

print("\n=== 2. NUMBER OF GIVEAWAYS AND AVERAGE SCORE BY YEAR ===")
yearly_stats = df_cleaned.groupby('Year').agg(
    Total_Games=('gameTitle', 'count'),
    Avg_Metacritic=('metacriticScore', 'mean')
).round(1)
print(yearly_stats)

print("\n=== 3. RATING CORRELATION (Critics vs Players) ===")
# Check correlation between different rating systems
correlation = df_cleaned[['epicRating', 'metacriticScore', 'steamDBRating']].corr().round(2)
print(correlation)

# --- VISUALIZATION ---
# Set the plot style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 5))

# Plot 1: Number of free games per year
plt.subplot(1, 2, 1)
sns.countplot(data=df_cleaned, x='Year', hue='Year', palette='viridis', legend=False)
plt.title('Number of Free Games in EGS by Year')
plt.xlabel('Year')
plt.ylabel('Number of Games')

# Plot 2: Metacritic vs SteamDB Rating correlation
plt.subplot(1, 2, 2)
sns.scatterplot(data=df_cleaned, x='metacriticScore', y='steamDBRating', alpha=0.6, color='b')
# Add a trend line
sns.regplot(data=df_cleaned, x='metacriticScore', y='steamDBRating', scatter=False, color='red')
plt.title('Correlation: Metacritic Score vs SteamDB Rating')
plt.xlabel('Metacritic Score')
plt.ylabel('SteamDB Rating (%)')

plt.tight_layout()

# Save the plots as a static image (requirement for the report)
plt.savefig('egs_analysis_plots.png', dpi=300)
plt.show()

print("\n✅ Plots successfully generated and saved to 'egs_analysis_plots.png'")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

print("=== Stage 4: Advanced Time-Series & Distribution Analysis ===")
df_cleaned['epicRating_100'] = df_cleaned['epicRating'] * 20
M = (df_cleaned[['metacriticScore']].median().round(2))
print(M)
# 1. Feature Engineering: Extract Month and Day of the Week
# Ми створюємо нові колонки з місяцем та днем тижня для глибшого аналізу
df_cleaned['Month'] = df_cleaned['freeDate'].dt.month_name()
df_cleaned['DayOfWeek'] = df_cleaned['freeDate'].dt.day_name()

# Set up the visualization grid (3 plots vertically)
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(3, 1, figsize=(14, 18))

# --- Plot 1: Seasonality (Games per Month) ---
months_order = ['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December']
sns.countplot(data=df_cleaned, x='Month', ax=axes[0], palette='magma',
              order=months_order, hue='Month', legend=False)
axes[0].set_title('Seasonality: Number of Free Games by Month', fontsize=14, fontweight='bold')
axes[0].set_xlabel('')
axes[0].set_ylabel('Number of Games')

# --- Plot 2: Day of the Week Analysis ---
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
sns.countplot(data=df_cleaned, x='DayOfWeek', ax=axes[1], palette='coolwarm',
              order=days_order, hue='DayOfWeek', legend=False)
axes[1].set_title('Giveaway Schedule: Free Games by Day of the Week', fontsize=14, fontweight='bold')
axes[1].set_xlabel('')
axes[1].set_ylabel('Number of Games')

# --- Plot 3: Boxplot of Metacritic Scores by Year (Finding Outliers) ---
sns.boxplot(data=df_cleaned, x='Month', y='metacriticScore', ax=axes[2],
            palette='Set2', order=months_order, hue='Month', legend=False)
axes[2].set_title('Quality Distribution: Metacritic Scores by Month (Boxplot)', fontsize=14, fontweight='bold')
axes[2].set_xlabel('Month')
axes[2].set_ylabel('Metacritic Score')
axes[2].axhline(y=float(M), linewidth=4, color='r')


plt.tight_layout()
plt.savefig('advanced_analytics.png', dpi=300)
plt.show()

print("✅ Advanced analytics plots successfully generated and saved!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

print("=== Stage 5: Feature Engineering (Quality Tiers Analysis) ===")

# 1. Створюємо логіку розподілу на категорії (Feature Engineering)
bins = [0, 59.9, 74.9, 89.9, 100]
labels = ['Poor (<60)', 'Average (60-74)', 'Good (75-89)', 'Masterpiece (90+)']

# Функція pd.cut автоматично розкидає ігри по цих 4 категоріях
df_cleaned['QualityTier'] = pd.cut(df_cleaned['metacriticScore'], bins=bins, labels=labels)

# 2. Виводимо текстову статистику для викладача
tier_counts = df_cleaned['QualityTier'].value_counts().sort_index()
print("\nNumber of games in each quality category:")
print(tier_counts)

# 3. Візуалізуємо результати
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Будуємо графік
sns.countplot(data=df_cleaned, x='QualityTier', palette='viridis', order=labels, hue='QualityTier', legend=False)
plt.title('Quality of Free Games: Metacritic Tiers', fontsize=15, fontweight='bold')
plt.xlabel('Quality Tier')
plt.ylabel('Number of Games')

plt.tight_layout()
plt.savefig('quality_tiers.png', dpi=300)
plt.show()

print("✅ Quality Tiers analysis completed!")
```
